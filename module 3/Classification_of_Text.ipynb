  - Labeled input;

- Inference phase:

  - X: Set of attributes or features ${x_1,x_2,...,x_n}$;

- When there are more than 2 possible classes => Multi-class Classification;
- When data instances can have 2 or more labels => Multi-label Classification.

## Questions to ask in Supervised Learning

# Identifying Features from Text

## Why is textual data unique?
- Textual data present an unique set of challenges;
- All the information you need is in the text;
- But features can be pulled out from text at different granularities;

## Types of textual features (I)
- Words:
  - By far the most common class of features;
  - Handling commonly-occurring words: Stop words ('the', 'a',...);
  - Normalization: Make lower case vs leave as-is;
  - Stemming/Lemmatization;

  - Character sub-sequences in words: "ing", "ion".


# Naive Bayes Classifiers
## Case study: Classifying text search queries
- Suppose you are interested in classifying search queries in 3 classes: Entertainment, Computer Science, Zoology;
- Most common class of the three is Enternainment;
- Suppose the query is "Python":
  - Python, the snake(Zoology);
  - Python, the programming language (Computer Science);
  - Python, as in Monty Python (Entertainment).

  - Most common class, given "Python", is Zoology.

  - Suppose the query is "Python download". Then the most probable class is Computer Science.

### Probabilistic model
- Update the likelihood of the class given new information;
- Prior Probability: Pr(y=Entertainment), Pr(y=CS), Pr(y=Zoology);
- Posterior probability: Pr(y=Entertainment|x="Python"),...

## Bayes' Rule
$Posterior\, probability = \frac{Prior\, probability\, *\, Likelihood}{Evidence}$

$Pr(y|x) = \frac{PR(y)\, *\, Pr(X|y)}{Pr(X)}$

## Naive Bayes Classification
- $Pr(CS|"Python")= \frac{PR(y=CS)\, *\, Pr("Python"|y=CS)}{Pr("Python")}$

- $Pr(Zoology|"Python")= \frac{PR(y=Zoology)\, *\, Pr("Python"|y=Zoology)}{Pr("Python")}$

- $Pr(CS|"Python")>Pr(Zoology|"Python") => y=CS$

- $y^* = argmax_y\, Pr(y|X) = argmax_y\, Pr(y)\, *\, Pr (X|y)$

### Naive assumption
- Given the class label, features are assumed to be independent of each other.
- $y^* = argmax_y\, Pr(y|X) = argmax_y\, Pr(y)\, *\, \prod_{i=1}^{n} Pr (X_i|y)$
- For the query "Python download" we have:
  $y^* = argmax_y\, Pr(y)\, *\, Pr("Python"|y)\, *\, Pr("download"|y)$

## Naive Bayes: What are the parameters
- Prior probabilities: Pr(y) for all y in Y;
- Likelihood: Pr($x_i$|y) for all features $x_i$ and labels y in Y
- If there are 3 clases (Y=3) and 100 features in X, how many parameters does Naive Bayes models have ?
R: Y + 2 * X * Y = 603

## Naive Bayes: Learning parameters
- Prior probabilities: Pr(y) for all y in Y:
  - Acquired from the training data;
  - Count the number of instances in each class;
  - If there are N instances in all, and n out of those are labeled as class y => Pr(y) = n/N

- Likelihood: Pr($x_i$|y) for all features $x_i$ and labels y in Y:
  - Count how many times features $x_i$ appears in instances labeled as class y;
  - If there are p instances of class y, and $x_i$ appears in k of those, Pr($x_i$|y)= k/p

## Naive Bayes: Smoothing
- What happens if Pr($x_i$|y) = 0 ?
  - Feature $x_i$ never occurs in documents labeled y;
  - But then, the posterior probability Pr(y|$x_i$) will be 0
- Instead, smooth the parameters;
- Laplace smoothing or Addictive smoothing: Add a dummy count:
  - Pr($x_i$|y)= (k+1)/(p+n)

# Support Vector Machines

## Decision Boundaries
- Classification function is represented by decision surfaces;
- Choosing a Decision Boundary:
  - Training data;
  - Caution with overfitting: Decision boundary learned over training data doesn't generalize to test data.
### Linear Boundaries:
- Characteristics:
  - Easy to find;
  - Easy to evaluate;
  - More generalizable: "Occam's razor"
- Finding a Linear Boundary:
  - Find the angular parameter w;
  - Many methods: Perceptron, Linear Least Squares, Linear Discriminative Analysis ...
  - Problem: If linearly separable, then infinite number of linear boundaries

## Support Vector Machines (SVM)
- Support Vector Machines are maximum-margin classifiers;
- They're linear classifiers that find a hyperplane to separate two classes of data: positive and negative;
- Given training data, SVMs finds a linear function w (weight vector):
  - $f(x_i) =\, <w.x_i> + b$
  - if $f(x_i)\geq 0, y_i = +1;else\,y_i = -1$

## SVM: Multi-class classification
- SVMs work only for binary classification problems;
- 2 different situations:
  - One vs Rest: For each class you train a different SVM that detect if an input is from a specific class or not.
  - One vs One: Train one SVM for each pair of classes. Total number of SVMs to be trained is a combinatory of n, 2 by 2.

## SVM Parameters (I): Parameter C
- Regularization: How much importance should you give individual data points as compared to better generalized model;
- Larger values of C = less regularization:
  - Fit training data as well as possible, every data point important;
- Small values of c = more regularization:
  - More tolerant to erros on individual data points.

## SVM Parameters (II): Other params
- Linear kernels usually work best for text data:
  - Other kernels include rbf, polynomial;
- multi_class: ovr(one-vs-rest);
- class_weight: Different classes can get different weights

````python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/Applied_Text_Mining/module 3/Amazon_Unlocked_Mobile.zip')

df.head()
````

````python
df = df.dropna()

# Removing neutral reviews
df = df[df['Rating'] != 3]

# Creating a column with positive reviews binary encoded
df['Positively Rated'] = np.where(df['Rating'] > 3, 1, 0)
df.head(10)
````

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['Reviews'],
                                                    df['Positively Rated'],
                                                    random_state=0)
````

### Question 4

Fit and transform the training data `X_train` using a Tfidf Vectorizer with default parameters. The transformed data will be a compressed sparse row matrix where the number of rows is the number of documents in `X_train`, the number of columns is the number of features found by the vectorizer in each document, and each value in the sparse matrix is the tf-idf value. First find the **max** tf-idf value for every feature.

What 20 features have the smallest tf-idf and what 20 have the largest tf-idf among the **max** tf-idf values?

Put these features in two series where each series is sorted by tf-idf value. The index of the series should be the feature name, and the data should be the tf-idf.

The series of 20 features with smallest tf-idfs should be sorted smallest tfidf first, the list of 20 features with largest tf-idfs should be sorted largest first. Any entries with identical tf-ids should appear in lexigraphically increasing order by their feature name in boh series. For example, if the features "a", "b", "c" had the tf-idfs 1.0, 0.5, 1.0 in the series with the largest tf-idfs, then they should occur in the returned result in the order "a", "c", "b" with values 1.0, 1.0, 0.5.

*This function should return a tuple of two series
`(smallest tf-idfs series, largest tf-idfs series)`.*

````python
from sklearn.feature_extraction.text import TfidfVectorizer

def answer_four():

    tfidf_vect = TfidfVectorizer().fit(X_train)
    X_train_transformed = tfidf_vect.transform(X_train)

    max_tfidfs = X_train_transformed.max(axis=0).toarray()[0]
    feature_names = np.array(tfidf_vect.get_feature_names_out())

    sorted_tfidf_indices = max_tfidfs.argsort()

    # Smallest 20 tf-idf values
    smallest_tfidfs_indices = sorted_tfidf_indices[:20]
    smallest_tfidfs_values = max_tfidfs[smallest_tfidfs_indices]
    smallest_tfidfs_features = feature_names[smallest_tfidfs_indices]

    smallest_tfidfs_series = pd.Series(smallest_tfidfs_values, index=smallest_tfidfs_features)

    # Largest 20 tf-idf values
    largest_tfidfs_indices = sorted_tfidf_indices[-20:][::-1] # Reverse to get largest first
    largest_tfidfs_values = max_tfidfs[largest_tfidfs_indices]
    largest_tfidfs_features = feature_names[largest_tfidfs_indices]

    largest_tfidfs_series = pd.Series(largest_tfidfs_values, index=largest_tfidfs_features)

    # Sorting them lexicographically
    smallest_tfidfs_series = smallest_tfidfs_series.sort_index().sort_values(kind='mergesort')
    largest_tfidfs_series = largest_tfidfs_series.sort_index().sort_values(kind='mergesort',ascending=False)
    return (smallest_tfidfs_series,largest_tfidfs_series)

answer_four()
````
